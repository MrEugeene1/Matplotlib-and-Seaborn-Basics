#### `LINE PLOT`

---

```python
# random data
x = np.linspace(0, 10, 1000)

# plot
plt.plot(x, np.cos(x), label='Cosine');
```

**📖 Explained simply**

- `np.linspace(0, 10, 1000)` → make **1,000 numbers evenly spaced** from 0 to 10 (0, 0.01, 0.02 … 10). These are the positions along the bottom ruler.
- `np.cos(x)` → for each of those numbers, calculate the **cosine** — a math wave that goes up and down between −1 and +1, like ocean waves 🌊.
- `plt.plot(x, y)` → draw a **line** through all the points. With 1,000 points so close together, it looks perfectly smooth.
- `label='Cosine'` → gives the line a name, but the name is only shown if you add `plt.legend()`.
- The `;` at the end hides a small technical text that Jupyter would otherwise print above the chart.

**How to read it:** the line goes **up and down regularly**, like waves. A line chart shows **how something changes** from left to right (usually over time).

👇 The same chart, done the complete way: with a title, axis names and the legend.

```python
x = np.linspace(0, 10, 1000)

plt.figure(figsize=(8, 3.5))
plt.plot(x, np.cos(x), label="Cosine")
plt.plot(x, np.sin(x), label="Sine")
plt.title("Two waves: cosine and sine")
plt.xlabel("x")
plt.ylabel("Value")
plt.legend()          # shows the names given with label=
plt.show()
```

#### `STACKED BAR PLOT`

---

```python
# sample dataframe
data = pd.DataFrame([ [1, 2],
                      [3, 4],
                      [8, 7],
                      [6, 5],
                      [4, 3],
                      [2, 1]])

data.plot(kind='bar', stacked=True);
```

**📖 Explained simply**

- `pd.DataFrame([...])` makes a **table** with **6 rows** and **2 columns**. The columns have no names, so pandas calls them **0** and **1**.

| row | column 0 | column 1 |
|---|---|---|
| 0 | 1 | 2 |
| 1 | 3 | 4 |
| 2 | 8 | 7 |
| 3 | 6 | 5 |
| 4 | 4 | 3 |
| 5 | 2 | 1 |

- `data.plot(kind='bar', stacked=True)` → pandas draws a **bar chart** for us (it uses Matplotlib inside).
- `stacked=True` → instead of two bars side by side, the two values are **piled on top of each other**, like blocks 🧱.

**How to read it:** each tower is **one row**. The bottom color is column 0, the top color is column 1.
The **total height** = column 0 + column 1. Row 2 is the tallest tower: 8 + 7 = **15**.

🧒 *Everyday example:* each tower is one day; the bottom block is money spent on food, the top block is money spent on transport. The total height is what you spent that day.

#### `Visualization With Seaborn`

---



---

- Seaborn is a Python data visualization library based on matplotlib.
- It provides a high-level interface for drawing attractive and informative statistical graphics. It provide choices for plot style and color defaults, defines simple high-level functions for common statistical plot types, and integrates with the functionality provided by Pandas DataFrames.
- The main idea of Seaborn is that it provides high-level commands to create a variety of plot types useful for statistical data exploration, and even some statistical model fitting.



```python
import seaborn as sns
sns.set()
```

**📖 Explained simply:** `import seaborn as sns` brings Seaborn (nickname `sns`).
`sns.set()` chooses a nicer **style** for all the following charts (grey background, grid lines).
In new versions of Seaborn the modern name is **`sns.set_theme()`** — it does the same thing.

> ⚠️ If your charts suddenly look different after reopening the notebook, you probably skipped this cell. Use **Run All**.

```python
# create random data
data = np.random.multivariate_normal([0, 0], [[5, 2], [2, 2]], size=2000)
data = pd.DataFrame(data, columns=['x', 'y'])
```

**📖 Explained simply**

This makes **pretend data**: **2,000 points**, each with an **x** and a **y** value, then puts them in a table with the column names `x` and `y`.

`np.random.multivariate_normal([0, 0], [[5, 2], [2, 2]], size=2000)` means:

| Piece | Meaning |
|---|---|
| `[0, 0]` | the **middle** (mean) of x is 0 and the middle of y is 0 |
| `[[5, 2], [2, 2]]` | the **spread** rules (the *covariance matrix*): **x spreads more** (5) than **y** (2), and the **2** in the corners means x and y **move together** a bit (when x is big, y tends to be big too) |
| `size=2000` | make 2,000 points |

⚠️ There's **no seed** here, so you get **different random numbers every time** you run the cell. To get the same numbers every time, put `np.random.seed(0)` on the line before.

#### `DENSITY  PLOTS`

---

```python
for col in 'xy':
    sns.kdeplot(data[col], shade=True)
```

**📖 Explained simply — the density plot (KDE)**

`for col in 'xy':` repeats the next line **twice**: once for column `x`, once for column `y`. So we get **two hills** on the same chart.

`sns.kdeplot(...)` draws a **KDE** (kernel density estimate): a **smooth hill** that shows **where most of the values are**.
🏖️ *Imagine every value pours a small cup of sand on a ruler. Where many values are close together, the sand piles up high. The KDE line is the top of the sand.*

**How to read it:**
- Both hills are **highest at 0**: most values are near 0 (we asked for a middle of 0).
- The **x hill is wider and lower** — x spreads more (5); the **y hill is narrower and taller** — y stays closer to 0 (2).
- The hills are **round and the same on both sides** → this shape is called a **normal distribution** or **bell curve** 🔔.

⚠️ `shade=True` is **old code**: new Seaborn shows a warning. The modern word is `fill=True` 👇

```python
for col in "xy":
    sns.kdeplot(data[col], fill=True, label=col)
plt.title("Where are most of the x and y values?")
plt.legend()
plt.show()
```

```python
for col in 'xy':
    sns.distplot(data[col],hist=False)
```

**📖 Explained simply:** `sns.distplot(..., hist=False)` draws the **same KDE hills** as above.

⚠️ `distplot` is **old** (*deprecated*): new Seaborn shows a warning and it will be removed one day.
The modern way to draw only the hill is **`sns.kdeplot`** 👇

```python
for col in "xy":
    sns.kdeplot(data[col], label=col)
plt.title("KDE hills, the modern way")
plt.legend()
plt.show()
```

```python
for col in 'xy':
    sns.distplot(data[col])
```

**📖 Explained simply:** `sns.distplot(...)` (without `hist=False`) draws **bars + the hill** together.

- The **bars** are a **histogram**: values are sorted into **jars** 🫙 by size. A tall bar = many values in that range.
- The **line** is the KDE: the smooth shape of the same data.
- Bars give the **facts**, the line gives the **shape** — using both is a good idea.

⚠️ `distplot` is **old**. The modern way is **`sns.histplot(..., kde=True)`** 👇

```python
for col in "xy":
    sns.histplot(data[col], kde=True, label=col, element="step")
plt.title("Histogram + KDE, the modern way")
plt.legend()
plt.show()
```

#### `PAIR PLOT`

---

```python
iris = sns.load_dataset("iris")
iris.head()
```

**📖 Explained simply — the iris dataset** 🌸

`sns.load_dataset("iris")` downloads a famous **practice table** about **150 iris flowers** (it needs an internet connection the first time).
`iris.head()` shows the **first 5 rows** — like peeping at the first page of a book.

| Column | Meaning |
|---|---|
| `sepal_length`, `sepal_width` | size of the **sepal** (the green leaf-like part under the petals), in cm |
| `petal_length`, `petal_width` | size of the **petal** (the colored part of the flower), in cm |
| `species` | the type of iris: **setosa**, **versicolor** or **virginica** (50 flowers each) |

The big question: *can we tell the 3 species apart just by measuring their flowers?*

***Visualizing the multidimensional relationships among the samples is as easy as calling sns.pairplot:***

```python
sns.pairplot(iris, hue='species', height=2.5);
```

**📖 Explained simply — the pair plot**

A pair plot compares **every column with every other column** in one picture. It's usually the **first chart** a data scientist draws on new data. 👋

- 4 number columns → 4 × 4 = **16 small charts**. `hue='species'` gives each species its own color. `height=2.5` = size of each small chart.
- ⛰️ **The diagonal** (top-left → bottom-right) shows **each column alone**, as KDE hills, one per species.
- ⚫ **All the other charts** show **two columns together**: each dot is **one flower**.
- 🪞 The top-right half is a **mirror** of the bottom-left half — read only one half.

**The tour (use this order for any pair plot):**
1. **Hills on the diagonal:** look at `petal_length` — the **setosa** hill is far away on its own (~1.5 cm), versicolor (~4.3 cm) and virginica (~5.6 cm) are further right. **Petal size separates the species very well.**
2. **Dots below the diagonal:** `petal_length` vs `petal_width` makes clear **stairs** 🪜 ↗️ — flowers with longer petals also have wider petals. Setosa makes its **own separate group** in the bottom-left corner.
3. **Colors:** setosa is always easy to separate; **versicolor and virginica overlap a little**, so they are harder to tell apart.

**What it means:** yes — just by measuring the **petals**, we can recognise the species almost every time (especially setosa). 🌸

> ⭐ "Diagonal = each thing alone (hills). Everything else = two things together (dots). Tour: hills → dots → colors."
